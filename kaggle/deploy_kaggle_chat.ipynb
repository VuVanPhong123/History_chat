```python
import os
import pickle
import numpy as np
import faiss
from sentence_transformers import SentenceTransformer, CrossEncoder
from rank_bm25 import BM25Okapi
import re
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM, StoppingCriteria, StoppingCriteriaList
from peft import PeftModel
import time
import gc
```

```python
class VietnameseRAGSystem:
    def __init__(self, embedding_model_name="intfloat/multilingual-e5-large-instruct", cross_encoder_name="namdp-ptit/ViRanker", cache_dir="/kaggle/working", device=None):
        if device is None:
            device = 'cuda' if torch.cuda.is_available() else 'cpu'
        self.device = device

        print("Loading embedding model with optimizations...")
        model_kwargs = {'torch_dtype': torch.float16} if device == 'cuda' else {}
        self.embedding_model = SentenceTransformer(
            embedding_model_name, 
            device=device,
            model_kwargs=model_kwargs,
            cache_folder=cache_dir
        )
        self.embedding_model.eval()
        if device == 'cuda':
            self.embedding_model.half()
        
        torch.cuda.empty_cache()

        print("Loading cross-encoder model with optimizations...")
        self.cross_encoder = CrossEncoder(cross_encoder_name, device=device)
        if device == 'cuda':
            self.cross_encoder.model.half()

        self.index = None
        self.chunks = []
        self.metadata = []
        self.cache_dir = cache_dir
        os.makedirs(cache_dir, exist_ok=True)
       
        self.retrieval_task = 'Given a query about Vietnamese history, retrieve relevant historical passages that answer the query'
        self.embedding_cache = {}
   
    def normalize_text(self, text):
        return text.lower()
   
    def get_detailed_instruct(self, query):
        return f'Instruct: {self.retrieval_task}\nQuery: {query}'  # Sửa lỗi \\n -> \n
   
    def load_index(self, faiss_file, metadata_file):
        # Sử dụng cache cho index
        cache_path = os.path.join(self.cache_dir, "index_cache.pkl")
        if os.path.exists(cache_path):
            print("Loading index from cache...")
            with open(cache_path, 'rb') as f:
                cached_data = pickle.load(f)
                self.index = cached_data['index']
                self.chunks = cached_data['chunks']
                self.metadata = cached_data['metadata']
                print(f"Loaded cached index with {len(self.chunks)} chunks")
                return
        
        # Tải index gốc
        self.index = faiss.read_index(faiss_file)
        with open(metadata_file, 'rb') as f:
            index_data = pickle.load(f)
       
        self.chunks = index_data['chunks']
        self.metadata = index_data['metadata']
       
        # Lưu cache
        with open(cache_path, 'wb') as f:
            pickle.dump({
                'index': self.index,
                'chunks': self.chunks,
                'metadata': self.metadata
            }, f)
       
        print(f"Loaded index with {len(self.chunks)} chunks")
   
    def hybrid_search(self, query, top_k=5, filter_trieu_dai=None, filter_chu_de=None):
        if self.index is None:
            raise ValueError("Index not initialized!")
       
        # Sử dụng cache cho kết quả tìm kiếm
        cache_key = f"{query}_{filter_trieu_dai}_{filter_chu_de}_{top_k}"
        if cache_key in self.embedding_cache:
            return self.embedding_cache[cache_key]
       
        initial_results = self._semantic_search(
            query,
            top_k=min(50, len(self.chunks)),
            filter_trieu_dai=filter_trieu_dai,
            filter_chu_de=filter_chu_de
        )
       
        if not initial_results:
            return []
       
        reranked_results = self._rerank_with_cross_encoder(query, initial_results, top_k=top_k)
        self.embedding_cache[cache_key] = reranked_results  # Lưu cache
        return reranked_results
    
    def _semantic_search(self, query, top_k=10, filter_trieu_dai=None, filter_chu_de=None):
        instructed_query = self.get_detailed_instruct(query)
        cache_key = f"embed_{instructed_query}"
        
        # Sử dụng cache cho embedding
        if cache_key in self.embedding_cache:
            query_embedding = self.embedding_cache[cache_key]
        else:
            with torch.no_grad():
                query_embedding = self.embedding_model.encode(
                    [instructed_query], 
                    normalize_embeddings=True,
                    convert_to_tensor=True
                ).cpu().numpy()
            self.embedding_cache[cache_key] = query_embedding
        
        candidate_size = min(top_k * 3, len(self.chunks))
        semantic_scores, semantic_indices = self.index.search(query_embedding, candidate_size)
        
        results = []
        for score, idx in zip(semantic_scores[0], semantic_indices[0]):
            if idx >= len(self.chunks) or score < 0.1:
                continue
                
            metadata = self.metadata[idx]
            
            # Áp dụng bộ lọc
            if filter_trieu_dai and metadata.get('trieu_dai') != filter_trieu_dai:
                continue
            if filter_chu_de and metadata.get('chu_de') != filter_chu_de:
                continue
            
            results.append({
                'text': self.chunks[idx]['text'],
                'metadata': metadata,
                'score': float(score),
                'original_index': int(idx)
            })
            
            if len(results) >= top_k:
                break
        
        return results
   
    def _rerank_with_cross_encoder(self, query, initial_results, top_k=5):
        if not initial_results:
            return []
       
        max_text_length = 384  # Giảm độ dài để tăng tốc
        documents = [result['text'][:max_text_length] for result in initial_results]
        pairs = [[query, doc] for doc in documents]
       
        try:
            # Tăng batch size và tắt progress bar
            ce_scores = self.cross_encoder.predict(
                pairs, 
                batch_size=64,  # Tăng từ 32 lên 64
                show_progress_bar=False,
                convert_to_tensor=True
            ).cpu().numpy()
        except Exception as e:
            print(f"Cross-encoder error: {e}. Use initial results.")
            return initial_results[:top_k]
       
        final_results = []
        for i, (ce_score, original_result) in enumerate(zip(ce_scores, initial_results)):
            semantic_score = original_result['score']
            combined_score = 0.65 * ce_score + 0.35 * semantic_score  # Điều chỉnh trọng số
           
            final_results.append({
                'text': original_result['text'],
                'metadata': original_result['metadata'],
                'score': float(combined_score),
                'ce_score': float(ce_score),
                'original_score': semantic_score
            })
       
        final_results.sort(key=lambda x: x['score'], reverse=True)
        return final_results[:top_k]
   
    def search(self, query, top_k=5, filter_trieu_dai=None, filter_chu_de=None):
        return self.hybrid_search(query, top_k, filter_trieu_dai, filter_chu_de)
   
    def get_available_filters(self):
        trieu_dais = set(m.get('trieu_dai') for m in self.metadata if m.get('trieu_dai'))
        chu_des = set(m.get('chu_de') for m in self.metadata if m.get('chu_de'))
       
        return {
            'trieu_dai': sorted(list(trieu_dais)),
            'chu_de': sorted(list(chu_des))
        }
```

```python
model_id = "Qwen/Qwen3-4B"
adapter_dir = "/kaggle/input/qwen-finetuned/transformers/default/3/qwen_finetuned"

os.environ["TRANSFORMERS_VERBOSITY"] = "error"

# Sử dụng tokenizer nhanh
tokenizer = AutoTokenizer.from_pretrained(
    adapter_dir,
    use_fast=True,  # Kích hoạt tokenizer nhanh
    padding_side="left"
)
tokenizer.pad_token = tokenizer.eos_token

base_model = AutoModelForCausalLM.from_pretrained(
    model_id,
    device_map="auto",
    torch_dtype=torch.float16,
    use_cache=True,
    low_cpu_mem_usage=True
)

model = PeftModel.from_pretrained(base_model, adapter_dir)
model = model.merge_and_unload()
model.eval()

# Kích hoạt JIT compilation cho GPU
if torch.cuda.is_available():
    try:
        model = torch.compile(model, mode="reduce-overhead", fullgraph=True)
        print("Model compiled with torch.compile for optimization")
    except Exception as e:
        print(f"Compilation failed: {e}. Using original model.")

if hasattr(model, 'config'):
    model.config.use_cache = True

print("Model and tokenizer loaded successfully")
print(f"Model device: {next(model.parameters()).device}")
```

```python
import re
import threading
import time
from transformers import TextIteratorStreamer
import gc
import torch
import requests
from bs4 import BeautifulSoup
from googleapiclient.discovery import build
from datetime import datetime

# --- CẤU HÌNH API (ĐIỀN KEY VÀO ĐÂY) ---
GOOGLE_API_KEY = ""
GOOGLE_CSE_ID = ""
MAX_CLASSIFICATION_TOKENS = 20
RETRIEVAL_THRESHOLD = 0.60

def clear_memory():
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()

class QuestionClassifier:
    def __init__(self, tokenizer, model, device):
        self.tokenizer = tokenizer
        self.model = model
        self.device = device
        self.classification_cache = {}
    
    def classify_question(self, question: str) -> str:
        print(f"[{datetime.now().strftime('%H:%M:%S')}] [CLASSIFIER] Bắt đầu phân loại câu hỏi: '{question[:50]}...'")
        
        if question in self.classification_cache:
            print(f"[{datetime.now().strftime('%H:%M:%S')}] [CLASSIFIER] Lấy từ cache: {self.classification_cache[question]}")
            return self.classification_cache[question]
        
        # ĐƠN GIẢN HÓA PROMPT - CHỈ ĐỂ PHÂN LOẠI
        classification_prompt = f"""Hãy phân loại câu hỏi sau. CHỈ TRẢ LỜI BẰNG MỘT TỪ DUY NHẤT: HISTORY_DIRECT, OUT_OF_DOMAIN, INSUFFICIENT_INFO, hoặc VAGUE.
    
    QUY TẮC:
    1. HISTORY_DIRECT: Câu hỏi về sự kiện, nhân vật, triều đại lịch sử Việt Nam. Kể cả câu hỏi ngắn gọn (có đối tượng rõ ràng ví dụ như nhân vật, bối cảnh, thời điểm, triều đại, ...)
    2. OUT_OF_DOMAIN: Câu hỏi HOÀN TOÀN KHÔNG LIÊN QUAN đến lịch sử/văn hóa Việt Nam (thời tiết, thể thao, ẩm thực, công nghệ hiện đại)
    3. INSUFFICIENT_INFO: Câu hỏi về lịch sử Việt Nam nhưng THÔNG TIN KHÔNG TỒN TẠI (ví dụ: công nghệ hiện đại trong thời phong kiến, phát minh không có thật)
    4. VAGUE: Câu hỏi QUÁ RỘNG, KHÔNG RÕ CHỦ NGỮ/VỊ NGỮ, hoặc không có từ khóa lịch sử cụ thể. Nếu đã có đối tượng cụ thể thì không phải là Vague (Ví dụ: "Kể đi", "Ông ấy là ai?", "Năm bao nhiêu?").
    
    VÍ DỤ:
    Q: "Vua Quang Trung đánh quân Thanh năm nào?" → HISTORY_DIRECT
    Q: "Thời tiết Hà Nội thế nào?" → OUT_OF_DOMAIN
    Q: "Ronaldo là ai?" → OUT_OF_DOMAIN
    Q: "Nhà Trần có dùng điện thoại không?" → INSUFFICIENT_INFO
    Q: "Kể về lịch sử Việt Nam" → VAGUE
    
    Câu hỏi cần phân loại: "{question}"
    Loại:"""
    
        print(f"[{datetime.now().strftime('%H:%M:%S')}] [CLASSIFIER] Prompt length: {len(classification_prompt)}")
        
        inputs = self.tokenizer(classification_prompt, return_tensors="pt", truncation=True, max_length=512)
        inputs = {k: v.to(self.device) for k, v in inputs.items()}
    
        try:
            print(f"[{datetime.now().strftime('%H:%M:%S')}] [CLASSIFIER] Đang chạy model để phân loại...")
            with torch.inference_mode():
                outputs = self.model.generate(
                    **inputs,
                    max_new_tokens=3,  # CHỈ CẦN 3 TOKEN (1 từ)
                    do_sample=False,
                    pad_token_id=self.tokenizer.eos_token_id,
                    eos_token_id=self.tokenizer.eos_token_id,
                    temperature=0.1,  # GIẢM temperature để ít ngẫu nhiên
                    repetition_penalty=1.2  # Tránh lặp lại
                )
    
            resp = self.tokenizer.decode(outputs[0], skip_special_tokens=True)
            
            # CHỈ LẤY PHẦN SAU "Loại:"
            if "Loại:" in resp:
                classification = resp.split("Loại:")[-1].strip()
            else:
                # Fallback: lấy phần cuối của response
                classification = resp.strip()
            
            print(f"[{datetime.now().strftime('%H:%M:%S')}] [CLASSIFIER] Raw response: '{resp}'")
            print(f"[{datetime.now().strftime('%H:%M:%S')}] [CLASSIFIER] Extracted: '{classification}'")
            
            # CHUẨN HÓA PHÂN LOẠI
            classification = classification.upper().strip()
            
            # LOẠI BỎ DẤU CÂU
            classification = classification.strip('":,.;!?')
            
            # KIỂM TRA TỪNG TRƯỜNG HỢP
            if "HISTORY_DIRECT" in classification or "HISTORY" in classification:
                result = "HISTORY_DIRECT"
            elif "OUT_OF_DOMAIN" in classification or "OUT" in classification or "DOMAIN" in classification:
                result = "OUT_OF_DOMAIN"
            elif "INSUFFICIENT_INFO" in classification or "INSUFFICIENT" in classification:
                result = "INSUFFICIENT_INFO"
            elif "VAGUE" in classification:
                result = "VAGUE"
            else:
                # FALLBACK: Kiểm tra từ khóa trong câu hỏi
                question_lower = question.lower()
                if any(word in question_lower for word in ["ronaldo", "messi", "bóng đá", "thể thao", "basketball", "football"]):
                    result = "OUT_OF_DOMAIN"
                elif any(word in question_lower for word in ["kể về", "là ai", "thế nào"]) and not any(word in question_lower for word in ["vua", "vị", "chiến", "trận", "triều"]):
                    result = "VAGUE"
                else:
                    result = "HISTORY_DIRECT"
                    
                print(f"[{datetime.now().strftime('%H:%M:%S')}] [CLASSIFIER] Fallback phân loại: {result}")
    
            self.classification_cache[question] = result
            print(f"[{datetime.now().strftime('%H:%M:%S')}] [CLASSIFIER] Phân loại thành công: {result}")
            return result
            
        finally:
            del inputs, outputs
            clear_memory()

class QASystemWithClassification:
    def __init__(self, tokenizer, model, device, rag_system):
        self.tokenizer = tokenizer
        self.model = model
        self.device = device
        self.rag_system = rag_system
        self.classifier = QuestionClassifier(tokenizer, model, device)
        
        self.model.eval()
        if hasattr(self.model, 'config'):
            self.model.config.use_cache = True
    
    def generate_answer(self, question: str):
        start_time = time.time()
        print(f"[{datetime.now().strftime('%H:%M:%S')}] [QA_SYSTEM] ========= BẮT ĐẦU XỬ LÝ CÂU HỎI =========")
        print(f"[{datetime.now().strftime('%H:%M:%S')}] [QA_SYSTEM] Nhận câu hỏi: '{question}'")
        
        try:
            question_type = self.classifier.classify_question(question)
            print(f"[{datetime.now().strftime('%H:%M:%S')}] [QA_SYSTEM] Loại câu hỏi đã phân loại: {question_type}")
            
            if question_type == "OUT_OF_DOMAIN":
                print(f"[{datetime.now().strftime('%H:%M:%S')}] [QA_SYSTEM] Xử lý câu hỏi NGOÀI PHẠM VI")
                yield from self._handle_out_of_domain_stream(question)
            elif question_type == "VAGUE":
                print(f"[{datetime.now().strftime('%H:%M:%S')}] [QA_SYSTEM] Xử lý câu hỏi MƠ HỒ")
                yield from self._handle_vague_question_stream(question)
            elif question_type == "INSUFFICIENT_INFO":
                print(f"[{datetime.now().strftime('%H:%M:%S')}] [QA_SYSTEM] Xử lý câu hỏi THIẾU THÔNG TIN")
                yield from self._handle_insufficient_info_stream(question)
            else:
                print(f"[{datetime.now().strftime('%H:%M:%S')}] [QA_SYSTEM] Xử lý câu hỏi LỊCH SỬ TRỰC TIẾP")
                yield from self._handle_history_question_stream(question)
                
            elapsed_time = time.time() - start_time
            print(f"[{datetime.now().strftime('%H:%M:%S')}] [QA_SYSTEM] ========= KẾT THÚC XỬ LÝ =========")
            print(f"[{datetime.now().strftime('%H:%M:%S')}] [QA_SYSTEM] Tổng thời gian xử lý: {elapsed_time:.2f} giây")
            
        finally:
            clear_memory()
    
    def _handle_out_of_domain_stream(self, question: str):
        print(f"[{datetime.now().strftime('%H:%M:%S')}] [OUT_OF_DOMAIN] Tạo prompt cho câu hỏi ngoài phạm vi")
        response_prompt = f"""Bạn là trợ lý chuyên gia về lịch sử Việt Nam. Hãy trả lời câu hỏi sau:
Câu hỏi: {question}
Đây là câu hỏi KHÔNG thuộc phạm vi lịch sử Việt Nam. Bạn cần:
1. Thông báo lịch sự rằng đây không phải câu hỏi về lịch sử Việt Nam
2. Giải thích ngắn gọn rằng bạn chỉ có thể hỗ trợ các câu hỏi về lịch sử Việt Nam
3. Mời người dùng hỏi về các chủ đề lịch sử Việt Nam khác
Hãy trả lời một cách chuyên nghiệp và lịch sự. Kết thúc bằng (end)
Trả lời:"""
        yield from self._generate_stream_core(response_prompt, "OUT_OF_DOMAIN")
    
    def _handle_vague_question_stream(self, question: str):
        print(f"[{datetime.now().strftime('%H:%M:%S')}] [VAGUE] Tạo prompt cho câu hỏi mơ hồ")
        response_prompt = f"""Bạn là trợ lý chuyên gia về lịch sử Việt Nam. Hãy trả lời câu hỏi sau:
Câu hỏi: {question}
Đây là câu hỏi QUÁ RỘNG và cần được LÀM RÕ. Bạn cần:
1. Nhận xét rằng câu hỏi này cần được cụ thể hóa
2. Đề nghị người dùng cung cấp thêm chi tiết về: thời gian, địa điểm, nhân vật hoặc sự kiện cụ thể
3. Đưa ra 1-2 ví dụ về cách hỏi cụ thể hơn
4. Giữ thái độ tích cực và hỗ trợ
Hãy trả lời một cách hữu ích. Kết thúc bằng (end)
Trả lời:"""
        yield from self._generate_stream_core(response_prompt, "VAGUE")
    
    def _handle_insufficient_info_stream(self, question: str):
        print(f"[{datetime.now().strftime('%H:%M:%S')}] [INSUFFICIENT_INFO] Bắt đầu xử lý câu hỏi thiếu thông tin")
        full_response = ""
        
        context = self._get_context_for_question(question)
        if context:
            print(f"[{datetime.now().strftime('%H:%M:%S')}] [INSUFFICIENT_INFO] Có context từ RAG (độ dài: {len(context)})")
            response_prompt = f"""Bạn là chuyên gia lịch sử Việt Nam. Dưới đây là dữ liệu trong hệ thống dữ liệu lịch sử của bạn:
{context}

Câu hỏi: {question}

Đây là câu hỏi về thông tin CÓ THỂ KHÔNG TỒN TẠI trong lịch sử. Bạn cần:
1. Phân tích tính hợp lý của câu hỏi trong bối cảnh lịch sử
2. Trả lời dựa trên dữ liệu có sẵn nếu có thông tin phù hợp
3. QUAN TRỌNG: Nếu dữ liệu tham khảo KHÔNG ĐỦ để khẳng định hoặc trả lời, hãy thừa nhận thẳng thắn, trong câu thừa nhận phải có: "Tôi không có đủ thông tin để trả lời câu hỏi này"
4. Cung cấp thông tin lịch sử liên quan nếu có thể
5. TUYỆT ĐỐI KHÔNG bịa đặt thông tin
Hãy trả lời một cách trung thực và chính xác. Kết thúc bằng (end)
Trả lời:"""
        else:
            print(f"[{datetime.now().strftime('%H:%M:%S')}] [INSUFFICIENT_INFO] Không có context từ RAG")
            response_prompt = f"""Bạn là chuyên gia lịch sử Việt Nam.
Câu hỏi: {question}

Đây là câu hỏi về thông tin CÓ THỂ KHÔNG TỒN TẠI trong lịch sử. Bạn cần:
1. Nếu không có đủ thông tin thì hãy thừa nhận thẳng thắn, trong câu bắt buộc phải có "Tôi không có đủ thông tin để trả lời câu hỏi này"
2. Không cố gắng bịa đặt câu trả lời.
Kết thúc bằng (end)
Trả lời:"""
        
        for chunk in self._generate_stream_core(response_prompt, "INSUFFICIENT_INFO"):
            full_response += chunk
            yield chunk
        
        failure_keywords = [
            "tôi không có đủ thông tin", 
            "không tìm thấy dữ liệu", 
            "không có thông tin trong hệ thống", 
            "chưa có thông tin", 
            "không đủ dữ liệu"
        ]
        is_failed = any(kw in full_response.lower() for kw in failure_keywords)
        
        if is_failed:
            print(f"[{datetime.now().strftime('%H:%M:%S')}] [INSUFFICIENT_INFO] Phát hiện từ khóa thất bại, tìm kiếm thêm trên internet")
            yield "\n\n(dữ liệu nội bộ không đủ -> Đang tìm kiếm thêm thông tin trên mạng...)\n"
            
            internet_context, source_links = self._search_google_api_and_get_context(question)
            
            if internet_context:
                print(f"[{datetime.now().strftime('%H:%M:%S')}] [INSUFFICIENT_INFO] Tìm thấy context từ internet (độ dài: {len(internet_context)})")
                response_prompt_internet = f"""Bạn là chuyên gia lịch sử Việt Nam. Dưới đây là DỮ LIỆU được truy xuất trong kho dữ liệu lịch sử của bạn:
{internet_context}

Câu hỏi: {question}

YÊU CẦU NGHIÊM NGẶT:
✅ Trả lời ĐẦY ĐỦ, CHÍNH XÁC dựa hoàn toàn trên tài liệu trên
✅ Tập trung vào thông tin trực tiếp liên quan đến câu hỏi
✅ Sắp xếp thông tin một cách logic và dễ hiểu
❌ Nếu tài liệu KHÔNG ĐỦ thông tin, hãy THỪA NHẬN RÕ RÀNG.
❌ TUYỆT ĐỐI KHÔNG thêm bất kỳ thông tin nào không có trong dữ liệu
Hãy trả lời bằng tiếng Việt có dấu đầy đủ. Kết thúc bằng (end)
Bắt đầu trả lời:"""
                
                for chunk in self._generate_stream_core(response_prompt_internet, "INSUFFICIENT_INFO_INTERNET"):
                    yield chunk
                
                if source_links:
                    yield f"\n\n**Nguồn tham khảo từ Internet:**\n{source_links}"
            else:
                print(f"[{datetime.now().strftime('%H:%M:%S')}] [INSUFFICIENT_INFO] Không tìm thấy context từ internet")
    
    def _handle_history_question_stream(self, question: str):
        print(f"[{datetime.now().strftime('%H:%M:%S')}] [HISTORY] Bắt đầu xử lý câu hỏi lịch sử")
        context = self._get_context_for_question(question)
        
        if context:
            print(f"[{datetime.now().strftime('%H:%M:%S')}] [HISTORY] Có context từ RAG (độ dài: {len(context)})")
            response_prompt = f"""Bạn là chuyên gia lịch sử Việt Nam. Dưới đây là DỮ LIỆU TRUY XUẤT ĐƯỢC từ hệ thống dữ liệu của chính bạn (chỉ sử dụng thông tin từ đây để trả lời):
{context}

Câu hỏi: {question}

YÊU CẦU NGHIÊM NGẶT:
✅ Trả lời ĐẦY ĐỦ, CHÍNH XÁC dựa hoàn toàn trên tài liệu trên, hãy bổ sung, cung cấp thêm một số thông tin liên quan(bổ sung thêm 3 đến 5 câu)
✅ Tập trung vào thông tin trực tiếp liên quan đến câu hỏi
✅ Sắp xếp thông tin một cách logic và dễ hiểu
✅ Có thể sử dụng định dạng để làm nổi bật thông tin quan trọng
❌ Nếu tài liệu KHÔNG ĐỦ thông tin, BẮT BUỘC phải thừa nhận thẳng thắn, trong phần thừa nhận bắt buộc phải có câu: "Tôi không có đủ thông tin để trả lời câu hỏi này"
❌ TUYỆT ĐỐI KHÔNG thêm bất kỳ thông tin nào không có trong dữ liệu
❌ KHÔNG sử dụng tiếng Anh trong câu trả lời
Hãy trả lời bằng tiếng Việt có dấu đầy đủ. Kết thúc bằng (end)
Bắt đầu trả lời:"""
        else:
            print(f"[{datetime.now().strftime('%H:%M:%S')}] [HISTORY] Không có context từ RAG")
            response_prompt = f"""Bạn là chuyên gia lịch sử Việt Nam.
Câu hỏi: {question}
THÔNG BÁO: Không tìm thấy dữ liệu phù hợp trong hệ thống để trả lời câu hỏi này.
Bạn cần:
1. Trong câu thông báo bắt buộc phải có câu: "Tôi không có đủ thông tin để trả lời câu hỏi này"
2. Đề xuất người dùng hỏi câu hỏi khác về lịch sử Việt Nam
3. Giữ thái độ hỗ trợ và chuyên nghiệp
Hãy trả lời một cách lịch sự. Kết thúc bằng (end)
Trả lời:"""
        
        full_response = ""
        print(f"[{datetime.now().strftime('%H:%M:%S')}] [HISTORY] Bắt đầu sinh câu trả lời")
        for chunk in self._generate_stream_core(response_prompt, "HISTORY"):
            full_response += chunk
            yield chunk
        
        print(f"[{datetime.now().strftime('%H:%M:%S')}] [HISTORY] Đã sinh xong câu trả lời (độ dài: {len(full_response)})")
        
        failure_keywords = [
            "tôi không có đủ thông tin", 
            "không tìm thấy dữ liệu", 
            "không có thông tin trong hệ thống", 
            "chưa có thông tin", 
            "không đủ dữ liệu"
        ]
        is_failed = any(kw in full_response.lower() for kw in failure_keywords)
        
        if is_failed:
            print(f"[{datetime.now().strftime('%H:%M:%S')}] [HISTORY] Phát hiện từ khóa thất bại, tìm kiếm thêm trên internet")
            yield "\n\n(Đang tìm kiếm thêm thông tin trên mạng...)\n"
            
            internet_context, source_links = self._search_google_api_and_get_context(question)
            
            if internet_context:
                print(f"[{datetime.now().strftime('%H:%M:%S')}] [HISTORY] Tìm thấy context từ internet (độ dài: {len(internet_context)})")
                response_prompt_internet = f"""Bạn là chuyên gia lịch sử Việt Nam. Dưới đây là DỮ LIỆU được truy xuất từ hệ thống dữ liệu của chính bạn:
{internet_context}

Câu hỏi: {question}

YÊU CẦU NGHIÊM NGẶT:
✅ Trả lời ĐẦY ĐỦ, CHÍNH XÁC dựa hoàn toàn trên tài liệu trên(Hãy bổ sung một số thông tin liên quan, bổ sung từ 5 đến 7 câu)
✅ Tập trung vào thông tin trực tiếp liên quan đến câu hỏi
✅ Sắp xếp thông tin một cách logic và dễ hiểu
✅ Có thể sử dụng định dạng để làm nổi bật thông tin quan trọng
❌ Nếu tài liệu KHÔNG ĐỦ thông tin, hãy THỪA NHẬN RÕ RÀNG, trong câu thừa nhận phải có câu "Tôi không có đủ thông tin để trả lời câu hỏi này"
❌ TUYỆT ĐỐI KHÔNG thêm bất kỳ thông tin nào không có trong dữ liệu
❌ KHÔNG sử dụng tiếng Anh trong câu trả lời
Hãy trả lời bằng tiếng Việt có dấu đầy đủ. Kết thúc bằng (end)
Bắt đầu trả lời:"""
                
                for chunk in self._generate_stream_core(response_prompt_internet, "HISTORY_INTERNET"):
                    yield chunk
                
                if source_links:
                    yield f"\n\n**Nguồn tham khảo từ Internet:**\n{source_links}"
            else:
                print(f"[{datetime.now().strftime('%H:%M:%S')}] [HISTORY] Không tìm thấy context từ internet")
    
    def _search_google_api_and_get_context(self, query: str, num_results=3):
        print(f"[{datetime.now().strftime('%H:%M:%S')}] [GOOGLE_SEARCH] Bắt đầu tìm kiếm Google cho: '{query}'")
        if "ĐIỀN_" in GOOGLE_API_KEY or "ĐIỀN_" in GOOGLE_CSE_ID:
            print(f"[{datetime.now().strftime('%H:%M:%S')}] [GOOGLE_SEARCH] ❌ LỖI: Bạn chưa điền GOOGLE_API_KEY hoặc GOOGLE_CSE_ID trong code!")
            return "", ""

        try:
            service = build("customsearch", "v1", developerKey=GOOGLE_API_KEY)
            search_query = f"{query} lịch sử Việt Nam"
            
            print(f"[{datetime.now().strftime('%H:%M:%S')}] [GOOGLE_SEARCH] Gửi request đến Google API: '{search_query}'")
            res = service.cse().list(
                q=search_query,
                cx=GOOGLE_CSE_ID,
                num=num_results
            ).execute()
            
            items = res.get('items', [])
            if not items:
                print(f"[{datetime.now().strftime('%H:%M:%S')}] [GOOGLE_SEARCH] Google API không trả về kết quả nào.")
                return "", ""

            print(f"[{datetime.now().strftime('%H:%M:%S')}] [GOOGLE_SEARCH] Nhận được {len(items)} kết quả từ Google")
            context_parts = []
            valid_links = []
            
            headers = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'}
            
            for i, item in enumerate(items, 1):
                url = item['link']
                print(f"[{datetime.now().strftime('%H:%M:%S')}] [GOOGLE_SEARCH] [{i}] Đang xử lý URL: {url}")
                try:
                    response = requests.get(url, headers=headers, timeout=5)
                    if response.status_code == 200:
                        soup = BeautifulSoup(response.text, 'html.parser')
                        
                        paragraphs = soup.find_all('p')
                        text_content = ' '.join([p.get_text() for p in paragraphs])
                        text_content = re.sub(r'\s+', ' ', text_content).strip()
                        
                        if len(text_content) > 100:
                            snippet = text_content[:1000] + "..."
                            context_parts.append(f"Nguồn ({url}): {snippet}")
                            valid_links.append(f"- {item['title']}: {url}")
                            print(f"[{datetime.now().strftime('%H:%M:%S')}] [GOOGLE_SEARCH] [{i}] Đã trích xuất {len(text_content)} ký tự")
                    else:
                        print(f"[{datetime.now().strftime('%H:%M:%S')}] [GOOGLE_SEARCH] [{i}] HTTP {response.status_code} cho URL {url}")
                except Exception as e:
                    print(f"[{datetime.now().strftime('%H:%M:%S')}] [GOOGLE_SEARCH] [{i}] Lỗi khi đọc link {url}: {e}")
                    continue
            
            full_context = "\n\n".join(context_parts)
            links_str = "\n".join(valid_links)
            
            print(f"[{datetime.now().strftime('%H:%M:%S')}] [GOOGLE_SEARCH] Hoàn tất, tổng cộng {len(context_parts)} nguồn hợp lệ")
            return full_context, links_str

        except Exception as e:
            print(f"[{datetime.now().strftime('%H:%M:%S')}] [GOOGLE_SEARCH] Lỗi Google API Search: {e}")
            return "", ""
    
    def _get_context_for_question(self, question: str, top_k: int = 4) -> str:
        print(f"[{datetime.now().strftime('%H:%M:%S')}] [RAG_RETRIEVAL] Bắt đầu retrieval cho: '{question}'")
        results = self.rag_system.search(question, top_k=min(20, top_k*2))
        
        if not results:
            print(f"[{datetime.now().strftime('%H:%M:%S')}] [RAG_RETRIEVAL] Không có kết quả trả về từ retrieval.")
            return None

        top_score = results[0].get('score', 0.0)
        print(f"[{datetime.now().strftime('%H:%M:%S')}] [RAG_RETRIEVAL] Điểm số top result: {top_score:.3f}")
        
        if top_score < RETRIEVAL_THRESHOLD:
            print(f"[{datetime.now().strftime('%H:%M:%S')}] [RAG_RETRIEVAL] Điểm số thấp hơn ngưỡng {RETRIEVAL_THRESHOLD}, không sử dụng context.")
            return None

        top_results = results[:min(4, len(results))]
        context_parts = []
        for i, result in enumerate(top_results, 1):
            text = result['text'].strip()
            score = result.get('score', 0.0)
            dynasty = result.get('metadata', {}).get('trieu_dai', 'Không rõ')
            print(f"[{datetime.now().strftime('%H:%M:%S')}] [RAG_RETRIEVAL] Kết quả {i}: Điểm={score:.3f}, Triều đại={dynasty}, Độ dài text={len(text)}")
            context_parts.append(text)

        total_length = sum(len(text) for text in context_parts)
        print(f"[{datetime.now().strftime('%H:%M:%S')}] [RAG_RETRIEVAL] Hoàn tất retrieval, tổng độ dài context: {total_length}")
        return "\n\n".join(context_parts)
    
    def _generate_stream_core(self, prompt: str, prompt_type: str = "GENERAL"):
        start_time = time.time()
        print(f"[{datetime.now().strftime('%H:%M:%S')}] [GENERATION][{prompt_type}] Bắt đầu generation (prompt độ dài: {len(prompt)})")
        
        inputs = None
        try:
            inputs = self.tokenizer(prompt, return_tensors="pt", truncation=True, max_length=2048)
            inputs = {k: v.to(self.device) for k, v in inputs.items()}

            streamer = TextIteratorStreamer(self.tokenizer, skip_prompt=True, skip_special_tokens=True, timeout=30)
            
            full_output = ""
            stop_generation = False
            
            def generate_with_streamer(inputs_dict):
                try:
                    print(f"[{datetime.now().strftime('%H:%M:%S')}] [GENERATION][{prompt_type}] Model generation bắt đầu")
                    with torch.inference_mode():
                        self.model.generate(
                            **inputs_dict,
                            max_new_tokens=1024,
                            do_sample=True,
                            temperature=0.3,
                            streamer=streamer,
                            pad_token_id=self.tokenizer.eos_token_id
                        )
                    print(f"[{datetime.now().strftime('%H:%M:%S')}] [GENERATION][{prompt_type}] Model generation hoàn tất")
                except Exception as e:
                    print(f"[{datetime.now().strftime('%H:%M:%S')}] [GENERATION][{prompt_type}] Generation error: {e}")
            
            generation_thread = threading.Thread(target=generate_with_streamer, args=(inputs,))
            generation_thread.start()
            
            chunk_count = 0
            for new_text in streamer:
                if stop_generation:
                    break
                
                chunk_count += 1
                full_output += new_text
                yield new_text
                
                if re.search(r'\(?end\)?', full_output, re.IGNORECASE):
                    stop_generation = True
                    print(f"[{datetime.now().strftime('%H:%M:%S')}] [GENERATION][{prompt_type}] Phát hiện token kết thúc (end)")
                    break
            
            generation_thread.join(timeout=5)
            
            elapsed_time = time.time() - start_time
            print(f"[{datetime.now().strftime('%H:%M:%S')}] [GENERATION][{prompt_type}] Generation hoàn tất")
            print(f"[{datetime.now().strftime('%H:%M:%S')}] [GENERATION][{prompt_type}] Thời gian: {elapsed_time:.2f}s, Số chunk: {chunk_count}, Độ dài output: {len(full_output)}")
            
        finally:
            if inputs is not None:
                del inputs
            clear_memory()
```

```python
import os
import socket
os.environ["TOKENIZERS_PARALLELISM"] = "false"
!pip install fastapi uvicorn python-multipart nest_asyncio pyngrok

import nest_asyncio
nest_asyncio.apply()

from fastapi import FastAPI, HTTPException
from fastapi.middleware.cors import CORSMiddleware
from fastapi.responses import StreamingResponse
from pydantic import BaseModel
import uvicorn
import threading
import time
import json
import asyncio
from typing import Optional
from pyngrok import ngrok, conf

NGROK_AUTH_TOKEN = os.environ.get("NGROK_AUTH_TOKEN", "")
if not NGROK_AUTH_TOKEN:
    print("Warning: NGROK_AUTH_TOKEN not configured.")
else:
    ngrok.set_auth_token(NGROK_AUTH_TOKEN)
    print("Ngrok authenticated")

app = FastAPI(
    title="History ChatBot API",
    description="AI ChatBot specialized in Vietnamese history", 
    version="1.0.0"
)

app.add_middleware(
    CORSMiddleware,
    allow_origins=["*"],
    allow_credentials=True,
    allow_methods=["*"],
    allow_headers=["*"],
)

class ChatRequest(BaseModel):
    question: str
    session_id: Optional[str] = None

qa_system_global = None

@app.on_event("startup")
async def startup_event():
    global qa_system_global
    print("Initializing QA System...")
    if 'qa_system' in globals():
        qa_system_global = qa_system
        print("QA System ready")
    else:
        print("ERROR: qa_system not found. Please run QA system initialization cell first.")

@app.get("/")
async def root():
    return {"message": "History ChatBot API is running"}

@app.get("/health")
async def health_check():
    return {
        "status": "healthy",
        "models_loaded": qa_system_global is not None,
    }

@app.post("/chat")
async def chat_endpoint(request: ChatRequest):
    if qa_system_global is None:
        raise HTTPException(status_code=503, detail="QA System not initialized")
    
    async def response_generator():
        try:
            for token in qa_system_global.generate_answer(request.question):
                if not token: continue
                
                clean_token = token.replace("(start)", "").replace("(Start)", "")\
                                   .replace("(end)", "").replace("(End)", "")
                
                if len(clean_token) == 0: continue

                chunk_data = {
                    "text": clean_token,
                    "session_id": request.session_id or "default",
                    "status": "generating",
                    "is_last": False
                }
                yield json.dumps(chunk_data, ensure_ascii=False) + "\n"
                await asyncio.sleep(0.001)
            
            yield json.dumps({
                "text": "",
                "session_id": request.session_id or "default",
                "status": "completed",
                "is_last": True
            }, ensure_ascii=False) + "\n"
            
        except Exception as e:
            error_data = {"error": str(e), "status": "error"}
            yield json.dumps(error_data) + "\n"

    return StreamingResponse(response_generator(), media_type="application/x-ndjson")

def get_free_port(start_port=8000):
    port = start_port
    while True:
        try:
            sock = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
            sock.bind(('0.0.0.0', port))
            sock.close()
            return port
        except OSError:
            port += 1

def start_ngrok_tunnel(port):
    try:
        conf.get_default().region = "us"
        ngrok.kill()
        time.sleep(1)
        
        tunnel = ngrok.connect(port, proto="http", bind_tls=True)
        print("=" * 60)
        print(f"PUBLIC HTTPS URL: {tunnel.public_url}")
        print("=" * 60)
        return tunnel
    except Exception as e:
        print(f"Ngrok error: {e}")
        return None

def run_server(port):
    config = uvicorn.Config(
        app, 
        host="0.0.0.0", 
        port=port, 
        log_level="critical",
        access_log=False,
        timeout_keep_alive=30,
    )
    server = uvicorn.Server(config)
    import asyncio
    asyncio.run(server.serve())

def start_server_with_ngrok():
    free_port = get_free_port(8000)
    print(f"Found free port: {free_port}")
    print(f"Starting FastAPI server on port {free_port}...")
    
    server_thread = threading.Thread(
        target=run_server, 
        args=(free_port,), 
        daemon=True,
        name="FastAPI-Server"
    )
    server_thread.start()
    time.sleep(3)
    
    tunnel = start_ngrok_tunnel(free_port)
    
    if tunnel:
        print(f"Streaming Endpoint: {tunnel.public_url}/chat")
        with open("ngrok_url.txt", "w") as f:
            f.write(tunnel.public_url)
    
    return server_thread, tunnel

def stop_ngrok_tunnel():
    try:
        ngrok.kill()
        print("Ngrok tunnel stopped")
    except:
        pass

if __name__ == "__main__":
    if 'qa_system' not in globals():
        print("ERROR: qa_system is not defined.")
    else:
        try:
            server_thread, tunnel = start_server_with_ngrok()
            
            if tunnel:
                print("\nServer is running. Press Stop to exit.")
                try:
                    while True:
                        time.sleep(1)
                except KeyboardInterrupt:
                    print("Stopping...")
            else:
                print("Failed to create tunnel")
                
        except Exception as e:
            print(f"Error: {e}")
        
        finally:
            stop_ngrok_tunnel()
```
